```python
from pyspark.sql.functions import current_timestamp,lit

Tables=["Customers","Geolocation","Order_Items","Order_Payments","Order_Reviews","Orders","Products","Sellers"]
```

```python
spark.sql("CREATE SCHEMA IF NOT EXISTS Olist_Lakehouse.bronze")
```

```python
for Table in Tables :
    df=spark.read.table(f"Olist_Lakehouse.raw.{Table}")
    df=df.withColumn("ingested_at",current_timestamp())\
            .withColumn("Source_Table",lit(Table))
    
    df.write.format("delta").mode("append").saveAsTable(f"Olist_Lakehouse.bronze.{Table}")        
```
