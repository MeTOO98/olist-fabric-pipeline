## Make connection to sql server

```python
import pyodbc
from dotenv import load_dotenv
import os
```

```python
load_dotenv()

server = os.getenv("DB_SERVER")
database = os.getenv("DB_DATABASE")
username = os.getenv("DB_USERNAME")
password = os.getenv("DB_PASSWORD")
driver = os.getenv("DB_DRIVER")
```

```python
conn_str = (
    f"DRIVER={{{driver}}};"
    f"SERVER={server};"
    f"DATABASE={database};"
    f"UID={username};"
    f"PWD={password};"
    f"TrustServerCertificate=yes;"
)
```

```python
conn = pyodbc.connect(conn_str)
cursor = conn.cursor()
```

## Now i Will make triggers to update last_modified_at when something changes in a row

```python
## Customers
cursor.execute("""
CREATE trigger tr_customers_modified
ON Customers
AFTER UPDATE
AS
BEGIN
    UPDATE C
    SET last_modified_at = GETDATE()
    FROM Customers C
    INNER JOIN inserted i ON C.customer_id = i.customer_id;
END;
 """)
conn.commit()
```

```python
## Orders
cursor.execute("""
CREATE trigger tr_orders_modified
ON Orders
AFTER UPDATE
AS
BEGIN
    UPDATE o
    SET last_modified_at = GETDATE()
    FROM Orders o
    INNER JOIN inserted i ON o.order_id = i.order_id;
END;
 """)
conn.commit()
```

```python
## Order_Items
cursor.execute("""
CREATE trigger tr_order_items_modified
ON Order_Items
AFTER UPDATE
AS
BEGIN
    UPDATE oi
    SET last_modified_at = GETDATE()
    FROM Order_Items oi
    INNER JOIN inserted i ON oi.order_id = i.order_id and oi.order_item_id = i.order_item_id;
END;
 """)
conn.commit()
```

```python
## Order_Payments
cursor.execute("""
CREATE trigger tr_order_payments_modified
ON Order_Payments
AFTER UPDATE
AS
BEGIN
    UPDATE op
    SET last_modified_at = GETDATE()
    FROM Order_Payments op
    INNER JOIN inserted i ON op.order_id = i.order_id and op.payment_sequential = i.payment_sequential;
END;
 """)
conn.commit()
```

```python
## Order_Reviews
cursor.execute("""
CREATE trigger tr_order_reviews_modified
ON Order_Reviews
AFTER UPDATE
AS
BEGIN
    UPDATE r
    SET last_modified_at = GETDATE()
    FROM Order_Reviews r
    INNER JOIN inserted i ON r.review_id = i.review_id ;
END;
 """)
conn.commit()
```
